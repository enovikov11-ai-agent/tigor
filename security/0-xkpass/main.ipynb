```python
from collections import Counter, defaultdict

with open("eff_large_wordlist.txt", "r") as file:
    lines = file.readlines()

words = [line.split("\t")[1].strip() for line in lines]

distribution = dict(Counter(len(word) for word in words))

counts = defaultdict(int)

for l1, c1 in distribution.items():
    for l2, c2 in distribution.items():
        for l3, c3 in distribution.items():
            for l4, c4 in distribution.items():
                counts[l1 + l2 + l3 + l4] += c1 * c2 * c3 * c4

size4_distribution = list(sorted(list(counts.items()), key=lambda x: x[0]))

total = sum(e[1] for e in size4_distribution)

[(l, c / total) for l, c in size4_distribution]
```
