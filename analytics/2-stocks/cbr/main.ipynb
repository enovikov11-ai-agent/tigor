```python
# view-source:https://www.cbr.ru/hd_base/keyrate/?UniDbQuery.Posted=True&UniDbQuery.From=17.09.2013&UniDbQuery.To=26.12.2025
import json

with open("data.json") as file:
    data = json.load(file)

dates = data["categories"]
day_rates = [(1 + (value / 100)) ** (1 / 365.25) for value in data["data"]]

dates[0], day_rates[0]
```

```python
from datetime import datetime, timedelta

start = datetime.strptime(dates[0], "%d.%m.%Y")
end = datetime.strptime(dates[-1], "%d.%m.%Y")

date_to_rate = {datetime.strptime(d, "%d.%m.%Y"): r for d, r in zip(dates, day_rates)}

filled_rates = []
current = start
prev_rate = day_rates[0]

while current <= end:
    if current in date_to_rate:
        prev_rate = date_to_rate[current]
    filled_rates.append(prev_rate)
    current += timedelta(days=1)

cumulative = 1
for rate in filled_rates:
    cumulative *= rate

cumulative
```
