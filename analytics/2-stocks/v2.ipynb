```python
import json

with open("v2.json", "r") as file:
    stocks_a = json.load(file)
    names = {s[0]: s[-1] for s in stocks_a}

with open("v2b.json", "r") as file:
    v2b = json.load(file)

stocks = [[e[0], float(e[1][0:-1]) * (1000 if e[1][-1] == "B" else 1), float(e[2]), names[e[0]]] for e in v2b if e[2] != ""]

stocks.sort(key=lambda x: x[2], reverse=True)
stocks = stocks[5:]

stocks.sort(key=lambda x: x[1], reverse=True)

total = sum(s[1] for s in stocks)
for s in stocks:
    s[1] = int(6265 * s[1] / total)

print(sum(s[1] for s in stocks))

stocks[0:12]
# ticker usd pe name
```
