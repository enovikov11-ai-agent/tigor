```python
from pathlib import Path
from random import shuffle
from PIL import Image

IMG_PATH = Path("in.png")   # change if the file lives elsewhere
GRID_N   = 16               # 16×16 grid → 256 tiles
```

```python
# ⬛️ 3. Load and auto-crop to 16-divisible dims
img = Image.open(IMG_PATH).convert("RGB")
w, h = img.size

# Trim overflow so both sides are multiples of 16
nw, nh = w - (w % GRID_N), h - (h % GRID_N)
if (nw, nh) != (w, h):
    img = img.crop((0, 0, nw, nh))
    w, h = nw, nh

tile_w, tile_h = w // GRID_N, h // GRID_N

img
```

```python
# ⬛️ 4. Slice into tiles
tiles = []
for row in range(img.size[1] // GRID_N):
    for col in range(img.size[0] // GRID_N):
        left   = col * tile_w
        upper  = row * tile_h
        right  = left + tile_w
        lower  = upper + tile_h
        tile = img.crop((left, upper, right, lower))
        tiles.append(tile)

print(f"{len(tiles)} tiles created.")
```

```python
# ⬛️ 5. Shuffle in-place
shuffle(tiles)
```

```python
# ⬛️ 6. (Optional) Peek at the first few shuffled tiles
#      This block stitches the first 4×4 shuffled tiles into a preview grid.

PREVIEW_N = 4
preview_canvas = Image.new("RGB", (tile_w * PREVIEW_N, tile_h * PREVIEW_N))
for idx, tile in enumerate(tiles[:PREVIEW_N**2]):
    r, c = divmod(idx, PREVIEW_N)
    preview_canvas.paste(tile, (c * tile_w, r * tile_h))

preview_canvas.show()  # Opens in the notebook viewer; comment out if headless
```
