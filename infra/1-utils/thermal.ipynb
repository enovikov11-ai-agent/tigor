```python
#!/usr/bin/env python3
from pathlib import Path
from math import ceil, sqrt
import argparse

from PIL import Image


def make_collage(input_dir: str | Path, output_png: str | Path) -> None:
    input_dir = Path(input_dir)
    output_png = Path(output_png)

    bmp_files = sorted(
        p for p in input_dir.iterdir()
        if p.is_file() and p.suffix.lower() == ".bmp"
    )

    if not bmp_files:
        raise ValueError(f"No BMP files found in {input_dir}")

    count = len(bmp_files)
    n = isqrt(count)

    n = ceil(sqrt(count))

    first = Image.open(bmp_files[0]).convert("RGB")
    tile_w, tile_h = first.size

    collage = Image.new(
        "RGB",
        (n * tile_w, n * tile_h),
        color=(0, 0, 0),  # black background
    )

    for idx, bmp_path in enumerate(bmp_files):
        with Image.open(bmp_path) as img:
            img = img.convert("RGB")

            if img.size != (tile_w, tile_h):
                raise ValueError(
                    f"{bmp_path} has size {img.size}, expected {(tile_w, tile_h)}"
                )

            x = (idx % n) * tile_w
            y = (idx // n) * tile_h

            # No scaling, exact paste.
            collage.paste(img, (x, y))

    collage.save(output_png, "PNG")


```
