```python
import torch
from diffusers import ZImagePipeline

pipe = ZImagePipeline.from_pretrained("/root/devbox-ssd/z-image-turbo", torch_dtype=torch.bfloat16, low_cpu_mem_usage=False)
pipe.to("cuda")
```

```python
prompt = "Two cute domestic cats, one orange tabby and one gray-and-white, sitting close together and happily munching on two computer RAM modules they are holding with their paws. The RAM looks edible but is clearly recognizable as DDR memory sticks with visible chips, circuit traces, and gold contacts. Warm soft daylight, shallow depth of field, cinematic bokeh, ultra detailed fluffy fur, adorable expressions, playful wholesome mood, high resolution, sharp focus."

image = pipe(
    prompt=prompt,
    height=400,
    width=400,
    num_inference_steps=50,
    guidance_scale=0.0,
    generator=torch.Generator("cuda").manual_seed(42)
).images[0]

image
```
