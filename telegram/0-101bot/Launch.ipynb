```python
session = await fetch("midjourney.com", "https://www.midjourney.com/api/auth/session/")

async def get_contents(page):
    return await fetch("midjourney.com", f"https://www.midjourney.com/api/app/recent-jobs/?page={page}&amount=50&orderBy=new&userId={session['user']['id']}")

async def clone():
    return_value = []
    i = 1
    result = await get_contents(i)
    while len(result) > 0:
        return_value.extend(result)
        i += 1
        result = await get_contents(i)
    return return_value

jobs = await clone()
filtered_jobs = {job['prompt']: job['image_paths'][0] for job in jobs if job['_job_type'] == "diffusion_upsample_v4"}
```
